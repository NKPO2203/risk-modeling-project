```python
import yfinance as yf
```

```python
symbole = "NVDA"

titre = yf.Ticker(symbole)

histo = titre.history(period = "max", auto_adjust = False)

print ("dimensions :", histo.shape)
print("colonnes :" , histo.columns.tolist())
```

```python
histo.head()
```

```python
histo.tail()
```

```python
divisions = histo[histo["Stock Splits"] !=0]
print("divisions d'actions :")
print(divisions[["Close", "Adj Close", "Stock Splits"]]) 
```

```python
dividendes = histo[histo["Dividends"] != 0]
print(len(dividendes) , "dividendes verses. Les cinq derniers : ")
print(dividendes[["Close" , "Adj Close", "Dividends"]].tail(5))
```

```python
derniere = divisions.index[-1]
position = histo.index.get_loc(derniere)
print()

print("les jours autour de la derniere divison: ")
print(histo.iloc[position-2 : position +2][["Close", "Adj Close" , "Stock Splits"]])
```
